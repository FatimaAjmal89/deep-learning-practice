```python
import torch
import torch.nn as nn
```

```python
torch.manual_seed(42)
inputs = torch.tensor([1.0,2.0,3.0])
```

```python
neurons = nn.Linear(in_features = 3,out_features = 1)
```

```python
output = neurons(inputs)
```

```python
output
```

```python
#verify using formula

neurons.weight @ inputs + neurons.bias
```
